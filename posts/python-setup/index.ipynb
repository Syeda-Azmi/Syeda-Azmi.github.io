## What are we doing today
In the last post we talked about what Python is, the tools around it, and why we'll be using uv for environment management. 
Now let's get everything installed and actually run some code. By the end of this post you'll have Python running, a notebook open in VS Code, and your first few lines of code executed.

## What so we need?

Here's everything this post sets up, in order:

1. **Python** — via uv, which installs and manages it for us
2. **uv** — our environment and package manager
3. **VS Code** — the editor we write and run code in
4. **Two VS Code extensions** — so VS Code understands notebooks

### Step 1 — Install VS Code

VS Code is the application you'll write all your code in. It's free, lightweight, and works well for everything from beginner notebooks to serious projects.

Download from: [code.visualstudio.com](https://code.visualstudio.com)

Click **Download for Windows**, run the installer, and keep all the default options. When it asks about adding VS Code to PATH — **check that box**. It'll save you trouble later.

### Step 2 — Install uv

uv is what we'll use to install Python and manage packages. Open the **Windows PowerShell** — search for it in the Start menu — and paste this command:

```powershell
powershell -ExecutionPolicy ByPass -c "irm https://astral.sh/uv/install.ps1 | iex"
```

Press Enter and let it finish. Once done, **close PowerShell and open a new one**. This is important — the new window picks up the changes the installer made.

Verify it worked by running:

```powershell
uv --version
```
*Note: You should see a version number printed. If you do, uv is installed*

> **What just happened?** That command downloaded and ran uv's installer script directly from the internet. The new PowerShell window is needed because your system's PATH — the list of places Windows looks for programs — only refreshes when a new terminal opens.

### Step 3 — Install Python via uv

Now let's use uv to install Python. Still in PowerShell, run:

```powershell
uv python install 3.12
```

This downloads and installs Python 3.12 — a stable, widely supported version. uv handles the whole thing; no need to visit python.org or run a separate installer.

Check it worked:

```powershell
uv run python --version
```

You should see `Python 3.12.x` printed.

> **Why uv and not python.org?** Both work. But since we're using uv to manage our projects anyway, letting it own the Python installation too keeps everything consistent and in one place.

### Step 4 — Install VS Code Extensions

Open VS Code. On the left sidebar you'll see a puzzle-piece icon — that's the **Extensions** panel. Click it and search for these two, installing both:

- **Python** — by Microsoft
- **Jupyter** — by Microsoft

These extensions are what allow VS Code to run Python code and open notebook files. Without them, VS Code is just a text editor.

### Step 5 — Create a Folder and Open Your First Notebook

Create a folder somewhere you'll remember — your Desktop or Documents is fine. Call it something like `python-blog` or `my-notebooks`.

In VS Code:
- Go to `File → Open Folder` and open that folder
- Click the **New File** icon in the file explorer panel
- Name it `first_notebook.ipynb`

The `.ipynb` extension is what tells VS Code this is a Jupyter notebook.

### Step 6 — Select a Kernel

When the notebook opens, you'll see a button in the top-right corner that says **Select Kernel**. Click it, then choose **Python Environments**, and select the Python 3.12 version that uv installed.

> **What is a kernel?** It's the engine that actually runs your code behind the scenes. When you click run on a cell, the kernel is what receives and executes it. Selecting the right one just means pointing your notebook at the Python installation you want to use.

## A Note on Environments — Do You Need One Right Now?

In the previous post we talked about virtual environments — isolated Python setups that keep projects separate from each other. The honest answer is: **not yet**.

Right now you have one Python installation and no projects pulling in conflicting libraries. For the next several posts where we're learning the basics, running directly from your uv-managed Python is perfectly fine.

Environments become important when:
- You're working on multiple projects that need different library versions
- You start installing data science libraries like pandas or PyTorch
- You want your setup to be reproducible — shareable with others or deployable somewhere

We'll set up a proper environment when we reach that point, and we'll walk through it in full when the time comes. For now, don't let it be a blocker.

## Let's Write Some Code

You should now have your notebook open with a kernel selected. Click the **+ Code** button to add a cell, and let's go.

### Cell 1 — Say Hello

````python
# This is a comment — Python ignores anything after the # symbol
# Comments are how you leave notes in your code, for yourself or others
# It's a good habit to annotate what your code is doing

print("Hello, Python!")
````

Press **Shift + Enter** to run the cell. The output appears directly below it, and the cursor moves to the next cell automatically. That's the core rhythm of working in notebooks — write in a cell, run it, see the result, move on.

### Cell 2 — Basic Math

````python
# Python can do arithmetic just like a calculator
print(10 + 3)   # Addition
print(10 - 3)   # Subtraction
print(10 * 3)   # Multiplication
print(10 / 3)   # Division
print(10 ** 2)  # Powers — this gives us 10 squared
print(10 % 3)   # Remainder — how much is left after dividing
````

Run this cell the same way. Notice each `print()` gives its own line of output. The `#` comments on each line are there to remind you what each operator does — this is exactly the kind of annotation that makes notebooks readable when you come back to them later.

> Tip: In a Jupyter notebook, you don't always need print() to see a result. If the last line of a cell is an expression — like 10 + 3 — the notebook will automatically display its output. Try it: delete the print() and just write 10 + 3 in a cell. However, this only works for values — writing "Hello, Python!" alone will display it, but it won't look or behave the same as print(), and plain text like Hello, Python! without quotes would just cause an error. For text output, print() is the reliable habit to keep.*

### Cell 3 — A Peek at What's Coming

````python
# What if we want to reuse a value instead of typing it every time?
# We can give it a name — this is called a variable
# We won't go deep on this yet, but try running this and see what happens

result = 10 + 3
print(result)
````

You don't need to fully understand this yet — we'll break down variables properly in the next post. But notice that Python stored the result of `10 + 3` and printed it when you asked. That idea — giving names to values — is one of the most fundamental things in programming, and it's what we're covering next.
