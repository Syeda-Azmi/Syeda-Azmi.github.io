## What is Python?
If you are here, you have probably heard about Python all around you but do not know where to start your journey. This is the right place. Begin with learning what it is, before learning how to do anything with it.

Python is a programming language — a way to give instructions to your computer in something closer to plain English. Unlike languages like C++ or Java, Python was designed to be readable and simple to write. That's why it became the go-to language for so many fields you are probably aware about including data science and AI.
Python has allowed programmers to focus more on where they are going with the code rather than focusing on writing the code itself.

## Why is Python used extensively?
There are various programming languages - some which are used in nichely. However, Python is used extensively, especially as a default in data science and machine learning. Why so?

- Python has a simple syntax allowing you to read and write the code almost like English sentences.
- Python is an interpreter, so code can be executed as soon as its written.
- It is free and open source. So you need no liscences or fees to be able to code.
- It has various pre-built toolkits, so that you do not have to build everything from scratch. 
- It is versatile, meaning you can use it not only for your dataset exploration and graph building but also for building models as well as a user interface. 

## Libraries
The pre-built toolkits I mentioned earlier? These are the libraries. Libraries are the most important aspect of Python, that makes it easy to use for variety of tasks. 

A library (also commonly called a package) is a collection of pre-written code that someone else built and very gratefully shared. Instead of writing hundreds of lines to load a spreadsheet, calculate statistics, or draw a chart — you import a library and call a function that does it in one line.

Think of it like the spices zou use to cook. You can gather everything and make spices from scratch at home but nobody does it because 'packaged' ones are convenient to use. 

Here are some libraries you will hear very often:

- numpy: core for scientific computing and handling arrays
- pandas: expert at handling data structures like tables
- matplotlib: handles your visualizations
- pytorch: your support system for everything neural network 

*side note: There are in-built and external libraries too. All the ones mentioned above are popular external ones. Check out the python documentation for a in-depth search at <https://docs.python.org/3/library/index.html>
or simply search about a library once you hear about it somewhere at <https://pypi.org/>*


## What is an IDE?
This is another term you will hear often once you enter the Python world. 

IDE stands for Integrated Development Environment. It's simply the app you use to write and run your code — think of it like Microsoft Word, but for programming. You can write your code there, edit, find errors, fix errors and run everything at the same place. It highlights your code, catches errors, and lets you run it all in one place. There are various IDEs available but we will be using VS Code throughout this blog. It's the most widely used editor, especially in the data science community right now and is also a good start for beginners as it allows you to shift to other languages too.

## Jupyter Notebook vs Python Script
Once you have an IDE, you'll write Python in one of two formats. This distinction trips up a lot of beginners, so it's worth clearing up early.

### Python Script — .py

A plain text file (ending with .py) containing Python code. When you run it, the entire file executes top to bottom. This is what you use for finished work — automation scripts, production code, tools you want to hand to someone else.

### Jupyter Notebook — .ipynb

A notebook where code lives in individual cells which you can run one at a time. Each cell shows its output directly below it. You can mix code cells with text cells — writing explanations, showing charts, and running code all in the same document. This is actually what is being used right now to write this blog. 

You can think of pyscript as a finished cookbook that you wrote for distribution whereas the notebook is your personal messy notebook with your experimentation and outcome breakdown. 

## Environment & Dependency Management
Here's where things get slightly confusing for beginners — and that's okay, because even experienced developers find this part messy. When you install Python, you also need to manage:

- Which version of Python you're using
- Which libraries you have installed
- Keeping different projects separate so their libraries don't conflict with each other

This is called environment management, and several tools exist to handle it. This is an essential part because it helps avoid conflicts — for example, one project might require PyTorch 2.x while another needs a different version. 

Let's walk through some of them.

| Tool | Function | Category |
|------|----------|----------|
| Anaconda | Manages Python, packages, environments | Distribution & environment management|
| pip | Install Python packages | Package Management |
| venv | Creates isolated python envrionments, uv is built on this concept | Environment management |
| uv | Manages Python versions, environments, packages, dependencies | **What we will be using mostly** |
