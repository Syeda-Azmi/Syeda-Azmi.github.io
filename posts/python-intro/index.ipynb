## What is Python?
Before learning to do anything with Python, we need to understand what even is it. Python is a programming language — a way to give instructions to your computer in something closer to plain English. Unlike languages like C++ or Java, Python was designed to be readable and simple to write. That's why it became the go-to language for data science, machine learning, and AI.
Python has allowed programmers to focus more on where they are going with the code rather than focusing on writing the code itself.

## Why is Python used extensively?
- Python has a simple syntax allowing you to read and write the code almost like English sentences.
- Python is an interpreter, so code can be executed as soon as its written.
- It is free and open source. So one needs no liscences or fees to be able to code.
- It has various pre-built toolkits including pandas (data handling), numpy (math), matplotlib (charts), and scikit-learn (machine learning), so that you do not have to build everything from scratch. 

## Some useful jargon
Before going further, here are terms you'll keep bumping into:
- Library / Package — A collection of pre-written code you can import and use. Like downloading an app instead of building one.
- Syntax — The grammar rules of a programming language. Get them wrong and the code breaks. This is similar to a sentence structure. If you do not follow a sentence structure, it does not mean anything!
- Output / Console — Where results of your code are printed/displayed.
- Variable — A named box that stores a value. (We will dive deeper into this in the next page.)
- Function — A reusable block of code that does a specific task. You call it by name. (This will be discussed further too.)

# What is an IDE?
This another term you will hear often once you enter the Python world. 
IDE stands for Integrated Development Environment. It's simply the app you use to write and run your code — think of it like Microsoft Word, but for programming. It highlights your code, catches errors, and lets you run it all in one place. There are different IDEs:
- VS code : Most commonly used. Used for general purpose
- PyCharm : Python specific and used for powerful, large projects

In my opinion, VS code is the best to use as a beginner who is not in any niche field as it is lightweight, free, has great extensions and allows you to shift to other languages if needed
