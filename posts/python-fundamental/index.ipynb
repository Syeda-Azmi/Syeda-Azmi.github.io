## Variables — Giving Names to Things

A variable is a name that points to a value. Instead of writing the same number or piece of text over and over, you store it once and refer to it by name.
> It is similar to calling a person by their name instead of referring to them by the description of their parents or ancestors

```python
age = 25
city = "Berlin"
temperature = 36.6
```

```python
age
```

That's it. The `=` sign here doesn't mean *equal to* — it means *store this value under this name*. You're making a label, not solving an equation.

You can reassign a variable at any time:

```python
age = 25
age = 26    # age is now 26, the old value is gone
```

```python
age
```

**Naming rules to know:**
- Names are case-sensitive — `age` and `Age` are two different variables
- Use lowercase with underscores for readability: `first_name`, not `FirstName` or `firstname`
- No spaces, no starting with a number, no special characters except `_`
> Learn a little more [here](https://peps.python.org/pep-0008/#naming-conventions)

```python
# Good names
user_age = 28
total_score = 150

# These will cause errors
2fast = 100       # starts with a number
my score = 99     # contains a space
```

## Data Types — What Kind of Value Is It?

Every value in Python has a type. Python figures this out automatically — you don't declare it — but knowing what type you're working with matters, especially when things don't behave as expected.
> You can think of this similar to peoples' personality types (MBTI/Zodiac signs etc). You dont always need to know what personality a person is but knowing it helps you work better with them. 

### Integers — `int`

Whole numbers, positive or negative, no decimal point.

```python
score = 42
temperature = -5
year = 2024
```

### Floats — `float`

Numbers with a decimal point.

```python
height = 1.82
pi = 3.14159
price = 9.99
```

> **Worth knowing:** Even `10.0` is a float, not an integer — the decimal point is what matters, not whether the decimal part is zero.

### Strings — `str`

Text. Anything inside quotes — single or double, both work.

```python
name = "Alice"
greeting = 'Hello, Python!'
```

Strings can hold anything — letters, numbers, symbols — but they're always *text*, not values you can do math with.

```python
# This works
print(10 + 5)       

# This breaks
print("10" + 5)     # Error — can't add a string and a number
```

### Booleans — `bool`

Only two possible values: `True` or `False`. Capitalisation matters.

```python
is_student = True
has_gpu = False
```

Booleans might seem trivial now but they're everywhere — every condition in a loop or an if-statement is producing a boolean behind the scenes.

## Checking and Converting Types

### `type()` — What type is this?

```python
x = 42
print(type(x))      

y = "hello"
print(type(y))      
```

You'll use `type()` constantly when debugging — especially when working with data that comes from files or the internet, where types aren't always what you expect.

### Converting Between Types

```python
# String to integer
age = int("28")
print(type(age))   

# Integer to float
score = float(10)
print(score)       

# Number to string
label = str(42)
print(label) 
print(type(label))         
```

> **When does this matter?** If you read data from a CSV file, numbers often arrive as strings. Trying to do math on them will break until you convert them. You'll run into this the moment you start using pandas.

## Operators

### Arithmetic Operators

You've seen these already in the last post — a quick reference:

```python
print(10 + 3)    
print(10 - 3)  
print(10 * 3)   
print(10 / 3)   
print(10 // 3)  
print(10 % 3) 
print(10 ** 2)  
```

### Comparison Operators

These return `True` or `False`. You'll need these the moment you write your first loop or condition.

```python
print(10 > 3)    
print(10 < 3)    
print(10 == 10)   # == checks equality and = assigns
print(10 != 5)    # not equal
print(10 >= 10)   
print(10 <= 9)    
```

> **Common mistake:** Using `=` when you mean `==`. A single `=` stores a value. A double `==` compares two values. Python will either throw an error or behave unexpectedly if you mix them up.

## Lists — Ordered Collections

So far every variable holds one value. A **list** holds multiple values, in order, in a single variable.

```python
temperatures = [22.1, 19.8, 25.3, 18.0, 23.5]
names = ["Alice", "Bob", "Carol"]
mixed = [42, "hello", True, 3.14]   # lists can hold different types
```

### Indexing — Accessing Items

Lists are ordered and each item has a position — called an **index** — starting from 0.

```python
names = ["Alice", "Bob", "Carol"]

print(names[0])    
print(names[1])   
print(names[-1])   #(negative index counts from the end)
```

### Basic List Operations

```python
scores = [85, 90, 78]

scores.append(95)       # add an item to the end
print(scores)           

print(len(scores))      # number of items in the list

print(scores[1:3])      # this is called slicing: items from index 1 up to (not including) 3
```

> **Slicing tip:** `[1:3]` gives you index 1 and 2 — the end number is always excluded. This feels odd at first but becomes natural quickly.

## Tuples — Like Lists, But Fixed

A tuple is similar to a list but **immutable** — once created, you can't change it.

```python
coordinates = (48.8566, 2.3522)   # latitude, longitude of Paris
print(coordinates[0])             
```

You'll mostly encounter tuples when functions return multiple values, or when working with data that shouldn't change — like coordinates or RGB colour values. You don't need to create them often, but you'll see them, so now you know what they are.

## Dictionaries — Key-Value Pairs

A dictionary stores data as **key-value pairs**. Instead of accessing items by position like a list, you access them by a meaningful name — the key.
> Similar to how language dictionaries work. You want to look up a meaning, you find the wrod in the dictionary

```python
person = {
    "name": "Alice",
    "age": 28,
    "city": "Berlin"
}

print(person["name"])   
print(person["age"])     
```

You can add or update entries:

```python
person["job"] = "engineer"      # adds a new key
person["age"] = 29              # updates an existing key
print(person)
```

> **Why does this matter later?** Dictionaries are everywhere in data work — JSON files (a common data format) are essentially dictionaries, and pandas works closely with dict-like structures. Getting comfortable with the key-value idea now pays off immediately.

## Lists vs Arrays — What's the Difference?

You'll hear the word **array** a lot once we start working with numpy. It's worth understanding how it differs from a list before we get there.

A Python **list** is flexible — it can hold different types, you can mix integers, strings, and booleans freely, and it's great for general-purpose use.

A **numpy array** is built for numerical computation. It holds one type of data, is significantly faster at math operations, and supports operations across the entire array at once rather than item by item.

```python
scores = [85, 90, 78]
scores * 2              # this just repeats the list, not multiplies each value

# With a numpy array, you can:
# scores * 2 would give [170, 180, 156], we will look further into this in the upcoming post
```

> Think of a list as a general-purpose container and a numpy array as a mathematical object. When you're storing and organising things, use a list. When you're doing computation across many numbers — which is almost everything in data science — you'll want arrays. We'll cover numpy properly in its own post.
