## If / Elif / Else — Making Decisions

A condition lets your code take different paths depending on whether something is true or false. This is the foundation of all logic in programming.

> It works exactly like how we carry out logic in our minds. If it rains, I will take the umbrella, otherwise I won't. That is the essence of the If-Elif-Else logic. 

```python
temperature = 35

if temperature > 30:
    print("It's hot outside")
```

If the condition (`temperature > 30`) is `True`, the indented block runs. If it's `False`, nothing happens.

> **Indentation matters in Python.** The code inside an `if` block must be indented — 4 spaces or one Tab. Python uses indentation to know what belongs inside the block. Get it wrong and you'll get an error. Try it!

### Adding `else`

```python
temperature = 22

if temperature > 30:
    print("It's hot outside")
else:
    print("It's not that hot")
```

`else` is the fallback — it runs when the condition is `False`.

### Adding `elif` — Multiple Conditions
`elif` stands for *else if*. It lets you check multiple conditions in sequence:

```python
temperature = 18

if temperature > 30:
    print("Hot")
elif temperature > 20:
    print("Warm")
elif temperature > 10:
    print("Cool")
else:
    print("Cold")
```

Python checks each condition top to bottom and runs the **first one that's true**, then skips the rest. Only one block ever runs.

### Combining Conditions

You can combine conditions with `and` / `or`:

```python
temperature = 28
humidity = 85

if temperature > 25 and humidity > 80:
    print("Hot and humid — not great")

if temperature > 35 or humidity > 90:
    print("Extreme conditions")
```

> - `and` — both conditions must be true
> - `or` — at least one must be true

## For Loops — Repeating Over a Sequence

A `for` loop goes through a collection — a list, a range of numbers, anything iterable — and runs the same block of code for each item.

```python
temperatures = [22.1, 19.8, 25.3, 18.0, 23.5]

for temp in temperatures:
    print(temp)
```

The variable `temp` takes on each value in the list one at a time. You choose the name — `temp`, `t`, `value` — it doesn't matter as long as you use the same name inside the block.

> Imagine you have a page with many shapes and you need to colour each one. It doesnt matter what colour you are using. You just need to color them. That's what it means by the name doesn't matter, what matters is whether you are being able to print it. 

### Using `range()`

When you want to repeat something a set number of times, `range()` generates a sequence of numbers for you:

```python
for i in range(5):
    print(i)      
```

`range(5)` gives you 0 through 4 — five numbers, starting at 0. You can also control the start, stop and step:

```python
for i in range(2, 10, 2):
    print(i)        
```

`range(start, stop, step)` — stop is always excluded.

### Combining a Loop with a Condition

This is where things start to get interesting — and this pattern appears constantly in data work:

```python
temperatures = [22.1, 35.8, 19.3, 38.1, 24.5]

for temp in temperatures:
    if temp > 30:
        print(f"{temp} is above threshold")
    else:
        print(f"{temp} is normal")
```

### `enumerate()` — When You Need the Index Too

Sometimes you want both the item and its position:

```python
temperatures = [22.1, 19.8, 25.3]

for index, temp in enumerate(temperatures):
    print(f"Day {index + 1}: {temp}°C")
```

> `enumerate()` is worth knowing early — you'll use it constantly when working with lists and later with dataframes.

## While Loops — Repeating Until a Condition Is Met

A `while` loop keeps running as long as a condition is `True`. Unlike a `for` loop where you know how many times it will run, a `while` loop runs an *unknown* number of times — until something changes.

```python
count = 0

while count < 5:
    print(f"Count is {count}")
    count += 1     
```

> **Infinite loops:** If the condition never becomes `False`, the loop runs forever and freezes your notebook. Always make sure something inside the loop is changing the condition. If you do get stuck, click the **Stop** button (■) in the toolbar to interrupt the kernel.

While loops are less common in data science than `for` loops, but they're useful when you're waiting for something — a model to converge, a value to reach a threshold, a process to complete.

## Break and Continue — Controlling Loop Behaviour

These two keywords give you finer control over what happens inside a loop.

### `break` — Exit the Loop Early

Stop the loop entirely when a condition is met, even if there are items left:

```python
temperatures = [22.1, 19.8, 35.8, 24.3, 38.1]

for temp in temperatures:
    if temp > 30:
        print(f"High value found: {temp} — stopping")
        break
    print(f"{temp} is normal")
```

Once `break` runs, the loop ends immediately. Everything after it in the list is ignored.

### `continue` — Skip to the Next Item

Skip the rest of the current iteration and move straight to the next one:

```python
temperatures = [22.1, None, 19.8, None, 25.3]

for temp in temperatures:
    if temp is None:
        continue        # skip missing values, keep going
    print(f"{temp}°C")
```

> **`continue` in data work:** Missing or invalid values are extremely common in real datasets. The pattern of checking for bad data and skipping it before processing is something you'll write constantly — and `continue` is often the cleanest way to do it.

## Mini Project — Analysing 2 Weeks of Temperature Readings

Let's put everything together. Imagine you've collected daily temperature readings for two weeks and want to do a basic analysis — flag anomalies, skip missing data, summarise what you found. Decide on your own rules. What are the values, what is an anomaly, what should the output look like. 

```python
# Two weeks of daily temperature readings (°C)
# None represents a missing reading — sensor failure, etc.
readings = [22.1, 19.8, 35.8, None, 23.5, 38.2, 21.0,
            None, 24.3, 20.1, 19.5, 41.0, 22.8, 36.5]

# Threshold above which we consider a reading anomalous
threshold = 30

# We'll collect valid readings and track anomalies
valid_readings = []
anomalies = []
missing_count = 0

# Loop through every reading
for index, temp in enumerate(readings):
    day = index + 1

    # Skip missing values
    if temp is None:
        print(f"Day {day}: No reading recorded")
        missing_count += 1
        continue

    # Flag anomalies
    if temp > threshold:
        print(f"Day {day}: {temp}°C — above threshold")
        anomalies.append(temp)
    else:
        print(f"Day {day}: {temp}°C — normal")

    # Collect all valid readings for summary
    valid_readings.append(temp)

# Summary
print("\n--- Summary ---")
print(f"Total days: {len(readings)}")
print(f"Missing readings: {missing_count}")
print(f"Valid readings: {len(valid_readings)}")
print(f"Anomalies detected: {len(anomalies)}")

# Only calculate average if we have valid data
if len(valid_readings) > 0:
    average = sum(valid_readings) / len(valid_readings)
    print(f"Average temperature: {average:.1f}°C")

if len(anomalies) > 0:
    print(f"Highest anomaly: {max(anomalies)}°C")
```

**What this uses:**
- A `for` loop with `enumerate()` to go through every reading with its day number
- `continue` to skip missing values cleanly
- `if / elif / else` to categorise each reading
- Two lists being built up as the loop runs — a pattern you'll see constantly in data processing
- A summary block using conditions to avoid errors on empty data

> **Notice the structure:** The loop handles one reading at a time. The summary happens after the loop finishes. This separation — process row by row, summarise at the end — is exactly how data pipelines work, just at much larger scale.


## Quick Reference

| Concept | Syntax | When to use |
|---|---|---|
| `if` | `if x > 0:` | Run code when condition is true |
| `elif` | `elif x == 0:` | Check another condition if previous was false |
| `else` | `else:` | Fallback when all conditions are false |
| `for` | `for item in list:` | Repeat over a known collection |
| `range()` | `range(start, stop, step)` | Generate a sequence of numbers |
| `enumerate()` | `for i, val in enumerate(list):` | Loop with index and value |
| `while` | `while condition:` | Repeat until condition becomes false |
| `break` | inside a loop | Exit loop immediately |
| `continue` | inside a loop | Skip current iteration, keep going |
