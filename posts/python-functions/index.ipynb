## What are we doing today
In this post we learn to write our own function. These are logic packaged into reusable and named blocks that lets us keep our code clean, readable and easy to fix (debug). Functions are one of the most important things in programming. And a good news, we have been using in-built functions already. Every time we've called `print()`, `len()`, or `type()`, we've been using a function.

## What is a Function?

A function is a named block of code that does a specific task. You define it once, and call it by name whenever you need it — once, ten times, inside a loop, anywhere.

Without functions:

```python
# Calculating a average for three different datasets
total1 = 22.1 + 19.8 + 25.3
average1 = total1 / 3

total2 = 30.5 + 28.1 + 33.2
average2 = total2 / 3

total3 = 15.0 + 18.4 + 20.1
average3 = total3 / 3

print (f"average1: {average1:.2f}, average2: {average2:.2f}, average3: {average3:.2f}")
```

With a function:

```python
def calculate_average(numbers):
    total = sum(numbers)
    return total / len(numbers)

average1 = calculate_average([22.1, 19.8, 25.3])
average2 = calculate_average([30.5, 28.1, 33.2])
average3 = calculate_average([15.0, 18.4, 20.1])

print (f"average1: {average1:.2f}, average2: {average2:.2f}, average3: {average3:.2f}")
```

Same result. Less repetition. And if something needs fixing, you fix it in one place.

## Anatomy of a Function

```python
def greet(name):
    message = f"Hello, {name}!"
    return message
```

- `def` — keyword that tells Python you're defining a function
- `greet` — the name you're giving it (same naming rules as variables)
- `(name)` — the **parameter**: a placeholder for the value you'll pass in
- `return` — sends a value back to whoever called the function

Calling it:

```python
result = greet("Syeda")
print(result)       
```

The value you pass in (`"Syeda"`) is called an **argument**. The placeholder inside the function (`name`) is the **parameter**. People use these words interchangeably — now you know both.

## Functions Without Parameters

Not every function needs input:

```python
def print_separator():
    print("-" * 40)

print_separator()
print("Some data here")
print_separator()
```

Useful for things you repeat often — formatting, headers, dividers.

## Functions Without Return

Not every function needs to return a value. Some just *do* something:

```python
def describe_temperature(temp):
    if temp > 30:
        print(f"{temp}°C — Hot")
    elif temp > 20:
        print(f"{temp}°C — Warm")
    else:
        print(f"{temp}°C — Cool")

describe_temperature(35)
describe_temperature(22)
describe_temperature(10)
```

> **Good to know:** If a function has no `return` statement, it returns `None` automatically. This is fine when you're using a function purely for its side effects — printing, saving a file, updating something.

## Multiple Parameters

Functions can take as many parameters as they need:

```python
def calculate_bmi(weight_kg, height_m):
    bmi = weight_kg / (height_m ** 2)
    return bmi

bmi = calculate_bmi(70, 1.75)
print(f"BMI: {bmi:.1f}")
```

## Default Parameter Values

You can give parameters a default value, making them optional when calling the function:

```python
def describe_reading(temp, unit="C"):
    print(f"Temperature: {temp}°{unit}")

describe_reading(22.5)          # uses default: °C
describe_reading(72.5, "F")     # overrides default: °F
```

Parameters with defaults must always come *after* parameters without defaults.

## Returning Multiple Values

A function can return more than one value — Python packs them into a tuple automatically:

```python
def analyse_readings(readings):
    average = sum(readings) / len(readings)
    maximum = max(readings)
    minimum = min(readings)
    return average, maximum, minimum

avg, high, low = analyse_readings([22.1, 35.8, 19.3, 38.1, 24.5])

print(f"Average: {avg:.1f}°C")
print(f"Highest: {high}°C")
print(f"Lowest:  {low}°C")
```

The three variables on the left unpack the three returned values in order. This is called **tuple unpacking** and you'll see it constantly.

## Functions Calling Other Functions

Functions can call other functions — this is how you build up complexity in layers while keeping each piece simple:

```python
def is_anomaly(temp, threshold=30):
    return temp > threshold

def classify_reading(temp):
    if temp is None:
        return "missing"
    elif is_anomaly(temp):
        return "anomaly"
    else:
        return "normal"

readings = [22.1, None, 35.8, 19.3, 38.1]

for temp in readings:
    status = classify_reading(temp)
    print(f"{temp} → {status}")
```

`classify_reading` delegates part of its job to `is_anomaly`. Each function stays focused on one thing. This is good practice — functions that do one thing well are easier to understand, test and fix.

> When functions start calling other functions, it would be good to skip the function definition and start from understanding the 'main' part of the code. And slowly trace back to see what the functions are doing. 

## Scope — Where Variables Live

Variables created inside a function only exist inside that function. They're not visible outside:

```python
def calculate_area(radius):
    pi = 3.14159          # this variable only exists inside this function
    area = pi * radius ** 2
    return area

print(calculate_area(5))
print(pi)                 # NameError — pi doesn't exist out here
```

This is called **scope**. It means functions are self-contained — they don't accidentally interfere with the rest of your code, and the rest of your code doesn't interfere with them.

> **The rule:** Variables defined inside a function are *local* to that function. Variables defined outside — at the top of your notebook — are *global* and can be read from anywhere, though modifying them from inside a function is something to be careful with.

> **Analogy:** Imagine these local and global variables like your fave celebrities. Globally known celebrities are known by everyone, everyone can access their information everywhere very easily. But a not so popular new artist from Bangladesh will be known by very niche people in Bangladesh. Information is also not available. 

## Mini Examples to Try

These get progressively more involved. Each one uses concepts from earlier posts.

**1. Celsius to Fahrenheit converter**

```python
def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32

print(celsius_to_fahrenheit(0))   
print(celsius_to_fahrenheit(100))
print(celsius_to_fahrenheit(22))  
```

**2. Count anomalies in a list**

```python
def count_anomalies(readings, threshold=30):
    count = 0
    for temp in readings:
        if temp is not None and temp > threshold:
            count += 1
    return count

data = [22.1, 35.8, None, 38.1, 24.5, 19.3]
print(count_anomalies(data))         
print(count_anomalies(data, threshold=20))   
```

**3. Clean a list of readings**

```python
def clean_readings(readings):
    # Remove None values and return only valid readings
    cleaned = []
    for temp in readings:
        if temp is not None:
            cleaned.append(temp)
    return cleaned

raw = [22.1, None, 35.8, None, 19.3]
clean = clean_readings(raw)
print(clean)   
```

**4. Full pipeline — clean, analyse, summarise**

```python
def clean_readings(readings):
    return [t for t in readings if t is not None]

def summarise(readings, threshold=30):
    if len(readings) == 0:
        print("No valid readings.")
        return

    average = sum(readings) / len(readings)
    anomalies = [t for t in readings if t > threshold]

    print(f"Readings:  {len(readings)}")
    print(f"Average:   {average:.1f}°C")
    print(f"Anomalies: {len(anomalies)} readings above {threshold}°C")
    if anomalies:
        print(f"Highest:   {max(anomalies)}°C")

raw_data = [22.1, None, 35.8, None, 38.1, 19.3, 24.5, None, 41.0]

cleaned = clean_readings(raw_data)
summarise(cleaned)
```

> **Notice** `clean_readings` uses a different syntax inside the brackets — `[t for t in readings if t is not None]`. This is called a **list comprehension** — a compact way to build a list with a loop and condition in one line. We won't go deep on it now, but it's worth recognising when you see it.

## Quick Reference

| Concept | Example | Notes |
|---|---|---|
| Define a function | `def my_func():` | Always followed by indented block |
| Parameter | `def f(x):` | Placeholder for input |
| Argument | `f(10)` | Actual value passed in |
| Return a value | `return x * 2` | Sends value back to caller |
| Default parameter | `def f(x=5):` | Optional when calling |
| Multiple returns | `return a, b, c` | Unpacked as `a, b, c = f()` |
| No return | function ends without `return` | Returns `None` automatically |
